```python
mydataset = {
    'cars': ["BMW", "Volvo", "Ford", "Tesla", "Bugatti", "VolksVagen", "KIA 5"],
    'passings': [3, 5, 7, 4, 7, 5, 8]
}

dataframe = pd.DataFrame(mydataset)
print(dataframe)
```

```python
dataframe.head(10) # shows a top list of items
```

```python
dataframe.tail() # show a reverse list of items
```

```python
dataframe.shape # used to determine the shape of tabular data
```

```python
dataframe.columns # used to identify the number of columns
```

```python
dataframe.dtypes # used to identify the type of tabular data features
```

```python
dataframe.describe() # used to describe the numerical columns
```

```python
# iloc() - used to retrieve data according to integer position
dataframe.iloc[0:5]
```

```python
# loc(2 args) - used to retrieve data from a table according to their int position and column feature names
dataframe.loc[1:5, "cars"]
```

```python
print(dataframe)
```

```python
dataframe[dataframe["passings"] !=4] # Conditional selection via operators
```

```python
# dataframe.mean(numeric_only=True)  - central value
dataframe["passings"].mean()
```

```python
# sum() - used to get a sum of all numeric and char values
dataframe["passings"].sum()
```
