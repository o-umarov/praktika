```python
import numpy as np

arr = np.array([2,5,6,8,10])
print("numpy array: ", arr)

arr2= np.array([[1,3,5,7], [2,4,6,8]])
print(f"numpy array: {arr2}")

print(f"arr1: {arr.shape}")
print(f"arr2: {arr2.shape}")
```

```python
a = np.array([1, 2, 3])
print(f"dType: {a.dtype}")
```

```python
# reshape() - change the shape of an array
a = np.array([1,2,3,4,5,6])
b = a.reshape(3,2)

print(f"Former: {a}")
print(f"Reshaped Array: {b}")
```

```python
# astype() - used to change the datatype
a = np.array([2.5, 4.8, 6.4, 1.8, 3.9, 5.7])
b = a.astype(int)
print(f"array a: {a} => {a.dtype}")
print(f"b array: {b} => {b.dtype}")
```

```python
# Dummy Data: zeros(); ones()

a = np.zeros((2,3))
b = np.ones((2,3))
print(f"a array: {a}")
print(f"b array: {b}")
```

```python
# ndim - used to identify array dimension
a = np.ones((3,4))
b = np.array([3,4,5])
print(f"a array dimension: {a.ndim}")
print(f"b array dimension: {b.ndim}")
```

```python
# arange - used to generate a sequence of array between a range of argument intervals

a = np.arange(1,10) # from 1 to 10 including 1 and excluding the last ele
print(f"a arange array: {a}")
```

```python
# sum() - used to calculate the sum of array eles
# min() - minimum
# max() - maximum
# mean() - between range
a = np.array([2,3,4,6,7,8,10])

print(f"a array sum: {a.sum()}")
print(f"a array mean: {(a.mean()):.2f}")
print(f"a array max: {a.max()}")
print(f"a array min: {a.min()}")
```

```python
# flatten() / ravel() - used to make 2D or 3D to 1D list array
a = np.array([[1,2], [3,4]])
print(f"flattened array: {a.flatten()}")
```

```python
# T() - Transpose - used to change shape (rearrange the array)
a = np.array([[1,4], [3,5]])
print(f"transposed arr: {a.T}")
```

```python
# dot() - multiply arrays (@)
a = np.array([[0,3], [2,4]])
b = np.array([[2,5], [1,6]])
print(f"a multiplied by b: {a.dot(b)}")
```

```python
# random.rand() - used to generate a random array
a = np.random.rand(2,3)
b = np.random.rand(2,3)
print(f"a random array:  {a}")
print(f"b random array:  {b} ")
```

```python
# linspace - Creates an array of evenly spaced numbers over a specified mathematical interval
a = np.linspace(2,3)
print(f"linspace array: {a}")
```
