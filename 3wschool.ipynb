```python
# string methods
print("omadi".upper())
print("OMADI".lower())
print("Hello Boss. ".strip())
print("Carlos Edisson".replace("Carlos", "Thomas"))
print("OMADDIYOR UMAROV ABDURAKHIM UGLI".split())
print("omaddiyor".capitalize())
print("OMADDIYOR".casefold())
print("Men ham ish qidirdim. Dostim ham ish qidirdim. Lekin ish chiqmadi".count("ish"))
print("Omaddiyor".encode())
print("I am here.".endswith("."))
print("I am here in Korea. Korea is good. I want to live here in Korea".index("Korea"))
```
