```python
# List Collection Methods
mylist = [2, 3, 5, 7, 11]
mylist[3:4] = [6]
print(mylist)
print(len(mylist))

# insert (index, item)
mylist.insert(0, 1) # add an item at the specific index
print(mylist)
print(len(mylist))

mylist.append(13) # add the item at the end
print((mylist))

odd_nums = [1, 3, 5, 7, 9]
even_nums = [2, 4, 6, 8, 10]
odd_nums.extend(even_nums)
print(odd_nums)

odd_nums.remove(1) # remove - delete the item according to var name
print(odd_nums)

odd_nums.pop(1) # pop - delete the item according to the index ** similar to del
print(odd_nums)

odd_nums.pop(len(odd_nums)-1)
print(odd_nums)

del odd_nums[0]
print(odd_nums)
print(even_nums)
del even_nums # in this line even_nums list get deleted from memory {Not available anymore}

# clear() - this method just clears all the list items making the list itself EMPTY with no items, lenght = 0
odd_nums.clear()
print(odd_nums) # []
```

```python
# Comprehension
# - Individual Task for myself: 1 dan 50 gacha bo'lgan sonlar ni bitta listga toplab, ular ichida 5 qatnashga barcha sonlarni ham bitta list ga toplab ularning medium qiymatini top.
nums50 = list(range(1, 51))
print(nums50)
numswith5 = [num5 for num5 in nums50 if str(5) in str(num5)]
print(numswith5)
sum = 0
for x in numswith5:
  sum+=x
result = float(sum/len(numswith5))
print(result)

randomlist = list(range(0,11))
print(randomlist)
new_random_with_odds = [y for y in randomlist if y%2==1]
print(new_random_with_odds)
```
